```python
import socket
```

```python
IP= '192.168.0.107'
PORT= 2023
ADDR=(IP,PORT)
SIZE=1024
FORMAT ='utf-8'
```

```python
#'file name' is the location path for the file to be stored
#for file receive from server
def main():
    client = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
    client.bind(("",PORT))
    client.connet(ADDR)
    print("CONNECTED TO SERVER")
    try:
        with open ('file name',"wb") as file:
            while True:
                file_data=client.recv(1024)
                if file_data:
                    file.write(file_data)
                else:
                    break
    except Exception as e:
        print('ERROR'.e)
    else:
        print('file name SUCCEED")
    client.close()
            
```

```python
main()
```
