```python
import socket
```

```python
IP= '192.168.0.107'
PORT= 1200
ADDR=(IP,PORT)
SIZE=1024
FORMAT ='utf-8'
```

```python
def main():
    client = socket.socket(socket.AF_INET,socket.SOCK_STREAM)
    client.connet(ADDR)
    print("[CONNECTED] Welcome to the server.")
    
    data=client.recv(SIZE).decode(FORMAT)
    print(f"[SERVER]{data}")
    
    data+="From CLIENT"
    client.send(data.encode(FORMAT))
    
    client.close()
```

```python
if __name__=="__main__":
    main
```
