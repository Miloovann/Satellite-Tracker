```python
from socket import *

host_name = "YY有名字的电脑"
port_num = 1555

clientSocket = socket(AF_INET, SOCK_STREAM)
clientSocket.connect((host_name,port_num))

message = input('enter a word:')
clientSocket.send(message.encode())

uppermessage = clientSocket.recv(1024).decode()
print(uppermessage)

clientSocket.close()
```
